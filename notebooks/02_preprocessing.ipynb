```python
import sys
import os
import pandas as pd

# 1. Modülleri bulabilmesi için bir üst dizini yola ekliyoruz
sys.path.append(os.path.abspath(".."))

from src.data_loader import DataLoader
from src.features import FeatureEngineer
from src.preprocessing import Preprocessor

# 2. Ham Verileri Yükle
loader = DataLoader(raw_data_dir="../data/raw", processed_data_dir="../data/processed")
train_df, test_df = loader.load_raw_data()

# 3. Feature Engineering Uygula
fe = FeatureEngineer()
train_fe = fe.create_features(train_df)
test_fe = fe.create_features(test_df)

# 4. Preprocessing Uygula (Fit sadece Train'de!)
prep = Preprocessor()
train_proc = prep.fit_transform(train_fe)
test_proc = prep.transform(test_fe)

# 5. Dönüştürülmüş Verileri data/processed/ Altına Kaydet
loader.save_processed_data(train_proc, test_proc)

print("\n--- SONUÇLAR ---")
print("İşlenmiş Train Veri Boyutu:", train_proc.shape)
print("İşlenmiş Test Veri Boyutu :", test_proc.shape)
print("\nİşlenmiş Train İlk 3 Satır:")
print(train_proc.head(3))
```
