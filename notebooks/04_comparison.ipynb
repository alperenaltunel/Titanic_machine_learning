```python
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Seaborn stil ayarları
sns.set_theme(style="whitegrid")
plt.rcParams['font.size'] = 11

# Oluşturduğumuz sonuçlar tablosunu okuyalım
results_df = pd.read_csv("../outputs/results_table.csv")

# F1-Score'a göre sıralayalım
results_df = results_df.sort_values(by="F1-Score", ascending=False).reset_index(drop=True)

print("Sonuç Tablosu Başarıyla Yüklendi! Toplam Model Sayısı:", len(results_df))
results_df[['Model', 'Val Accuracy', 'F1-Score', 'ROC-AUC', 'Fit Time (sec)']].head(10)
```

```python
os.makedirs("../outputs/figures", exist_ok=True)

fig, axes = plt.subplots(1, 2, figsize=(16, 6))

# 1. F1-Score Karşılaştırma Grafiği
sns.barplot(data=results_df, y='Model', x='F1-Score', palette='viridis', ax=axes[0])
axes[0].set_title('Modellerin F1-Score Performansı', fontsize=14, fontweight='bold')
axes[0].set_xlim(0.6, 0.85)
for p in axes[0].patches:
    axes[0].annotate(f"{p.get_width():.4f}", 
                     (p.get_width() - 0.04, p.get_y() + p.get_height() / 2.), 
                     ha='center', va='center', color='white', fontweight='bold', fontsize=9)

# 2. ROC-AUC Karşılaştırma Grafiği
sns.barplot(data=results_df, y='Model', x='ROC-AUC', palette='magma', ax=axes[1])
axes[1].set_title('Modellerin ROC-AUC Performansı', fontsize=14, fontweight='bold')
axes[1].set_xlim(0.7, 0.9)
for p in axes[1].patches:
    axes[1].annotate(f"{p.get_width():.4f}", 
                     (p.get_width() - 0.035, p.get_y() + p.get_height() / 2.), 
                     ha='center', va='center', color='white', fontweight='bold', fontsize=9)

plt.tight_layout()
plt.savefig("../outputs/figures/model_comparison.png", dpi=300)
plt.show()

print("[GÖRSEL KAYDEDİLDİ] outputs/figures/model_comparison.png")
```
